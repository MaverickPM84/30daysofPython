```python
# python weight converter

weight = float(input("ENter your weight: "))

unit = input("Kilograms or Pounds (K / P)")

if unit == "K":
    weight *= 2.205
    print(f" You are FAT, Your weight is {round(weight,2)}  in pounds")

elif unit == "P":
    weight /= 2.205
    print(f" You weight is {round(weight,2)} in Kgs")

else:
    print("Enter a valid unit")


```
