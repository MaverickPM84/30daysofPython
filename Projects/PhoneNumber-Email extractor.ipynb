Create a phone number, email id extractor using regular expressions

import re

# Text to search
text = "Hello, my email is example@example.com"

# Define a pattern to match email addresses
pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"

# Search for the pattern in the text
match = re.search(pattern, text)

# Check if a match is found
if match:
    email = match.group()
    print("Found email:", email)
else:
    print("No email found.")


MetaCharacters are helpful, significant and will be used in module re functions, which helps us comprehend the analogy with RE. The list of metacharacters is shown below.

\ sed to drop the special meaning of character following it.
[] Represent a character class.
^ Matches the beginning.
$ Matches the end.
. Matches any character except newline.
| Means OR (Matches with any of the characters separated by it.
? Matches zero or one occurrence.
* Any number of occurrences (including 0 occurrences).
+ One or more occurrences.
{} Indicate the number of occurrences of a preceding RegEx to match.
() Enclose a group of RegEx.

```python
import re

text = "My mobile number is 45507919 and email id - @gmail.com"

#define a pattern to match

mobile_pattern = r"(?:\+91[-.\s]?)?[6-9]\d{9}"
email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"

# Search for the pattern in the text
match_mobile = re.search(mobile_pattern, text)
match_email = re.search(email_pattern, text)


# Check if a mobile match is found
if match_mobile:
    mobile_number = match_mobile.group()
    print("Found mobile:", mobile_number)
else:
    print("No mobile number found.")

# check if email match is found

if match_email:
    email = match_email.group()
    print("Found email:", email)
else:
    print("Email not found")
```

```python
#refactored code


import re
from dataclasses import dataclass

# ── Patterns ────────────────────────────────────────────
PATTERNS = {
    "mobile": r"(?:\+91[-.\s]?)?[6-9]\d{9}",
    "email":  r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
    "url": r"https?://[^\s]+"
}

# ── Extractor ───────────────────────────────────────────
def extract_matches(text: str, patterns: dict[str, str]) -> dict[str, list[str]]:
    """Return all regex matches for each named pattern."""
    return {
        name: re.findall(pattern, text)
        for name, pattern in patterns.items()
    }

def display_results(results: dict[str, list[str]]) -> None:
    """Pretty-print extraction results."""
    for name, matches in results.items():
        if matches:
            label = name.replace("_", " ").title()
            print(f"Found {label}(s): {', '.join(matches)}")
        else:
            print(f"No {name} found.")

# ── Run ─────────────────────────────────────────────────
text = "My mobile number is +91-8645507919 and email id - pk@gmail.com and my website is https://aipmstudio.com/articles"

results = extract_matches(text, PATTERNS)
display_results(results)
```
