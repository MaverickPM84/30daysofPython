```python
# A value flowing through an API pipeline — same name, three different types
response = '{"role": "assistant", "content": "Hi"}'   # str: raw text off the wire
print(response)
print(type(response))

import json
response = json.loads(response)    # now a dict — tag re-tied to a new object
print(response)
print(type(response))

response = response["content"]     # now a str again — just the message text
print(response)
print(type(response))
```

```python
greeting = "hello"
print(id(greeting))    # identity: a fixed address in memory
print(type(greeting))  # type: what kind of thing it is
print(greeting)        # value: the data it holds
```

```python
x = 42
print(id(x))      # the int object 42 lives at this address

x = "hi"          # re-tie the tag x to a DIFFERENT object (a str)
print(id(x))      # different address = different object

x = [1, 2, 3]     # re-tie again to yet ANOTHER object (a list)
print(id(x))      # different address again
print(x)
```

```python
name = "Sam" 
print(name)
print(id(name))

lower_case = name.lower()
print(lower_case)
print(id(name.lower()))
```

```python
msgs = ["hi"]
msgs.append("bye")        # correct — edits in place, msgs is now ["hi", "bye"]
print(msgs)
msgs = msgs.append("x")   # BUG — append returns None, msgs is now None
print(msgs)
```

```python
# messages is a LIST (mutable) — we add turns in place.
messages = [{"role": "system", "content": "You are a helpful assistant."}]
messages.append({"role": "user", "content": "Summarize this article."})
messages.append({"role": "assistant", "content": "Sure—here's a summary..."})

print(len(messages)) 
print(messages)        # how many turns?
print(messages[1]["role"])   # who's the 2nd turn?
```

```python
#str immutable, returns new object

raw = "   Explain RAG in simple terms.\n"

raw.strip()           # makes a NEW string, but we threw it away
print(repr(raw))      # original still has the spaces + newline

clean = raw.strip().lower()   # capture the new string
print(repr(clean))
```
