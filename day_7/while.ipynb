#while loop, run this loop while this condition is true

```python
name = input("Enter your name: ")

while name == "":
    print("you did not enter your name")
    name = input("Enter your name: ")

print(f"Hello {name}")

```

```python
food = input ("Enter your fav food (q to quit)")

while not food == "q":
    print(f"you like {food}")
    food = input("What else do you like?(q to quit): ")

print("Bye")
```

```python
principal = 0
rate = 0
time = 0

while principal <= 0:
    principal = float(input("Enter the principal amount: "))
    
    if principal <= 0:
        print("Principal cannot be less than or equal to 0")


print(f"Principal amount is {principal}")

while rate <= 0:
    rate = float(input("Enter the interest rate: "))
    
    if principal <= 0:
        print("interest rate cannot be less than or equal to 0")


print(f"interest rate  is {rate} percent")


while time <= 0:
    time = int(input("Enter the loan period: "))
    
    if principal <= 0:
        print("loan period cannot be less than or equal to 0")

print(f"loan period  is {time} years")

final_amount = principal * pow((1 + rate/100),time)

print(f"The final amount after {time} years is {final_amount: .2f}")

```
