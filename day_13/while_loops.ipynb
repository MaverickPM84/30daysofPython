```python
#while loop - use case PIN 

print('BANK OF CODÉDEX')

pin = int(input('Enter your PIN: '))

while pin != 1234:
  pin = int(input('Incorrect PIN. Enter your PIN again: '))

if pin == 1234:
  print('PIN accepted!')
```

```python
#while loop will run over and over until the condition is true.
#Note: If the condition is False from the get-go, then the code block wouldn't run at all and will be skipped.

# This will run over and over again until the user guesses the number 6:
guess = 0

while guess != 6:
  guess = int(input('Guess the number: '))
  
print("correct guess")
```

```python
guess = 0
tries = 0

while guess != 6 and tries != 3:
  guess = int(input("Guess the number:  "))
  tries+=1
     
if guess != 6:
  print("Guess limit reached")
else:
  print("You got it")
```
