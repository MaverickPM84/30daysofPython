# We have created an object employee1 with class Employee, but we need data like - name, age, department

Where do we store these ?

Inside something called the constructor -
__init__()

It simply means, "Whenever someone create a new object, run this function first:

```python
class Employee:

    def __init__(self):
        print("Creating a new employee")

employee1 = Employee()
```

Python automatically called __init__().

You never called it yourself.

Answer:

Because of self.

Think of self as meaning:

"Me" or "This object."

Storing Values
class Employee:

    def __init__(self, name, age):

        self.name = name
        self.age = age

Let's trace this slowly.

You write

employee1 = Employee("Preetam", 32)

Python secretly does something like

Create new Employee object

↓

Run __init__()

↓

name = "Preetam"

age = 32

↓

Store inside object

Memory becomes

employee1

name → "Preetam"

age → 32

Now create another.

employee2 = Employee("Rahul", 28)

Memory

employee2

name → Rahul

age → 28

Each object has its own data.

Why self.name = name?

This line confuses everyone initially.

self.name = name

The left side

self.name

means

Store inside the object.

The right side

name

means

The parameter that was passed in.

Example

Employee("Preetam",32)

↓

name = "Preetam"

↓

self.name = "Preetam"

So the object permanently remembers it.

Accessing Data

Now we can do

print(employee1.name)

Output

Preetam

Or

print(employee2.age)

Output

28

Notice

We never created

employee1_name

Everything stays organized inside the object.

Objects Can Have Methods Too

Suppose every employee can introduce themselves.

class Employee:

    def __init__(self, name):
        self.name = name

    def introduce(self):
        print("Hi, I'm", self.name)

Now

employee1 = Employee("Preetam")

employee1.introduce()

Output

Hi, I'm Preetam

Create another.

employee2 = Employee("Rahul")

employee2.introduce()

Output

Hi, I'm Rahul

Same method.

Different object.

Different result.

```python
class Employee:

    def __init__(self, name):
        self.name = name
        
    def introduce(self):
        print(f"Hello, My name is {self.name}")

employee1 = Employee("Preetam")

employee1.introduce()
```

```python
class Book:

    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

book1 = Book("atomic Habits", "James Clear", 550)

book2 = Book("Deep Work", "Cal Newport", 480)

print(book1.title)

print(book2.author)

print(book1.price)
```
