```python
# 🌡️ 3. Weekly Temperature Tracker - Scenario: You record daily temperatures and want a summary.

temperature = [20, 23, 26, 30, 18, 19, 21]

highest_temp = max(temperature)

print(highest_temp)

lowest_temp = min(temperature)

print(lowest_temp)

average_temp = round(sum(temperature) / len(temperature), 2)

print(average_temp)
```

```python
# 5. Top 3 Scores in a Game - Scenario: Display a leaderboard from a list of scores.

scores = [450, 350, 400, 320, 475]

scores.sort(reverse=True) # Highest First
top_3 = scores[:3]


print("Leaderboard: ")

for top in top_3:
    print("-", top)
```

```python
scores = [450, 230, 890, 120, 670, 340]

scores.sort(reverse=True)  # Highest first
top_3 = scores[:3]         # Slice the top 3

print("🏆 Leaderboard:")
for i, score in enumerate(top_3, start=1):
    print(f"  #{i}: {score} points")
```
