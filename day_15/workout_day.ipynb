```python
#workout

print("#######################################")

workout = ['Back', 'Chest']
back_exercises = ['Pull ups', 'Rows', 'Back Extension']
chest_workout = ['Bench Press', 'Pec Dec', 'Incline Bench Press']


for workout in workout:
    print(workout)
    for exercise in back_exercises:
        print(exercise)
        for set in range(1, 4):
            print(f"Set{set}")
            for rep in range(1,11):
                print(f"Rep{rep}")
    print("#########################Back workout done, Chest Workout Starts########################################################")
    for exercise in chest_workout:
        print(exercise)
        for set in range(1,4):
            print(f"Set{set}")
            for rep in range(1,11):
                print(f"Rep{rep}")
```
