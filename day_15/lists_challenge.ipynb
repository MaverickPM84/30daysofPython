Are you feeling lucky? Let's replicate a small lottery with Python lists and the random module. 🤑

Create empty lists my_numbers & winning_numbers.

Loop over range(0,5), populate both lists with five random numbers between 1-69.

Outside the loop, add to both lists one more random number between 1-26.

Print the my_numbers and winning_numbers lists.

```python
import random

my_numbers = []
winning_numbers = []

for i in range(0,5):
    my_numbers.append(random.randint(1,69))
    winning_numbers.append(random.randint(1,69))

rand_num = random.randint(1,26)

my_numbers.append(rand_num)

rand_num_1 = random.randint(1,26)
winning_numbers.append(rand_num_1)

print(my_numbers)
print(winning_numbers)
```

Suppose you and three friends go out for a meal. The bill comes, and you decide to split it evenly. 👯👯

Here are the prices of what was ordered:

bill = [13.99, 28.75, 9.99, 9.99, 6.95, 7.45, 16.45, 16.45]

Using this bill list, calculate the total amount and then divide it four ways.

Initialize the total with zero and loop through the bill list to add everything up.

With the total, store what each person has to pay in a my_share variable and then print the result.

```python
total = 0

bill = [13.99, 28.75, 9.99, 9.99, 6.95, 7.45, 16.45, 16.45]

for i in range(0, len(bill)):
    total += bill[i]
    
print(total)
my_share = total / 4

print(my_share)


```

Instructions
A DNA sequence, such as 'ACG', TCA','CGT', is a three-letter combination of these pieces:

Adenine (A)
Cytosine (C)
Guanine (G)
Thymine (T)

Let's use a Python list to find a specific three-letter combination, starting with the following:

dna_sequence = ['GCT', 'AGC', 'AGG', 'TAA', 'ACT', 'CAT', 'TAT', 'CCC', 'ACG', 'GAA', 'ACC', 'GGA']

Next, define two variables:

item_to_find string with a three-letter combination of your choice (e.g. 'TGA').
item_found boolean, initialized to False.
Loop through each item in the dna_sequence list. Inside, use an if statement to see if an item is equal to item_to_find. If so, set item_found to True.

Outside the loop, use an if/else statement to check if item_found is True:

If so, print something like "Item Found!"
Else, print something like "Item not found."

```python
dna_sequence = ['GCT', 'AGC', 'AGG', 'TAA', 'ACT', 'CAT', 'TAT', 'CCC', 'ACG', 'GAA', 'ACC', 'GGA']

item_to_find = 'ACT'

item_found = False

for item in dna_sequence:
    if item_to_find == item:
      item_found = True
   
if item_found:
    print("Item Found")
else:
    print("Item not found")
```
