Large Language Models (shortened to LLMs) process and generate text. They’re trained on massive amounts of data and work by predicting the next token. But what's a token?

# How AI Reads
Humans read words. AI doesn't. Instead, AI breaks text into smaller chunks.

Tokens are the small chunks of text that AI models read and generate. 🪙

Sometimes a token is a whole word. Sometimes it's part of one. Sometimes it's punctuation.

token diagram

## Example 1
Here's how an AI model might see a sentence with punctuation: Beep beep boop.

['Beep', 'beep', 'boop', '.']

The AI breaks the sentence into four tokens. The period is its own token.

## Example 2
Here’s a sentence with a contraction: We’re so back!

['We', '’', 're', 'so', 'back', '!']

Six tokens. AI breaks We’re into smaller pieces.

So, how do we turn text into tokens?

# Simple Tokenizer
To create a simple tokenizer, we can use regular expressions.

Regular expression (shortened to regex) is a tool for describing the shape of text we want to extract. Instead of checking characters one by one, we can tell Python what text to find.

First, we need to import Python's built-in regular expression re module:

import re

For creating tokens, we want either a word, or a piece of punctuation. The regex for that is:

\w+|[^\w\s]

\w+: one or more word characters
|: or
[^\w\s]: a single character that is neither a word character nor whitespace
Simply put, look for a word or a punctuation mark.

Then, we can use re.findall() to go through a text and return the snippets found in a list:

text = 'The group chat is unhinged.'

re.findall(r'\w+|[^\w\s]', text)

If we print out the last line, it would look like:

['The', 'group', 'chat', 'is', 'unhinged', '.']

Don't sweat the syntax. You don't need to memorize regex to follow along here.

# Real World
Different models split words differently; there's no one-size-fits-all approach.

Models like GPT and Claude even use a fancy approach called subword tokenization. Long, rare, or unfamiliar words get broken into smaller, reusable pieces that can stand alone or help form other words. So doomscrolling might become ['doom', 'scroll', 'ing'].

This is why LLMs can handle words they've never seen before.

```python
import re

def tokenize(text):
    tokens = re.findall(r'\w+|[^\w\s]', text)
    return tokens

print(tokenize("Hi How are you?"))
print(tokenize('I’m gonna be King of the Pirates!'))
print(tokenize('Learn 👏 how 👏 AI 👏 works 👏'))
```
