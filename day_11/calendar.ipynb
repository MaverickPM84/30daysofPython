The `calendar` module is surprisingly feature-rich! Here's what you can do with it:

### 📅 Display Calendars

```python
import calendar

# Single month (what you're already doing)
print(calendar.month(2026, 6))

# Full year calendar
print(calendar.calendar(2026))

# HTML versions (great for web apps)
hc = calendar.HTMLCalendar()
print(hc.formatmonth(2026, 6))     # month as HTML table
print(hc.formatyear(2026))         # full year as HTML
```

### 🔢 Get Data (not just display)

```python
# Days of a month as nested list (weeks)
calendar.monthcalendar(2026, 6)
# → [[1,2,3,4,5,6,7], [8,9,...], ...]  (0 = day belongs to another month)

# Day range of a month → (starting_weekday, num_days)
calendar.monthrange(2026, 6)       # → (0, 30)  Monday start, 30 days

# What day of the week is a date?
calendar.weekday(2026, 6, 21)      # → 6 (Sunday)
```

### 📆 Year & Leap Year Utilities

```python
calendar.isleap(2026)              # → False
calendar.leapdays(2000, 2026)      # → 7 (leap years between range)
```

### 🗓️ Day & Month Names

```python
calendar.day_name[:]               # → ['Monday', ..., 'Sunday']
calendar.day_abbr[:]               # → ['Mon', ..., 'Sun']
calendar.month_name[1:]            # → ['January', ..., 'December']
calendar.month_abbr[1:]            # → ['Jan', ..., 'Dec']
```

### ⚙️ Customize Start of Week

```python
# Default week starts on Monday (0). Change to Sunday:
calendar.setfirstweekday(calendar.SUNDAY)
print(calendar.month(2026, 6))     # Now Sunday is the first column
```

### 🧱 OOP Style (TextCalendar / HTMLCalendar)

```python
# TextCalendar — fine-grained control
tc = calendar.TextCalendar(firstweekday=6)  # Start on Sunday
tc.prmonth(2026, 6)                         # Print month
tc.pryear(2026)                             # Print year

# Iterate over days (great for building custom UIs)
for day in tc.itermonthdays(2026, 6):
    print(day)  # 0 = padding day from adjacent month
```

### Quick Reference

| Function | Returns |
|---|---|
| `month(y, m)` | Formatted month string |
| `calendar(y)` | Formatted year string |
| `monthcalendar(y, m)` | Nested list of weeks |
| `monthrange(y, m)` | `(first_weekday, num_days)` |
| `weekday(y, m, d)` | Day of week (0=Mon, 6=Sun) |
| `isleap(y)` | `True` / `False` |
| `leapdays(y1, y2)` | Count of leap years in range |
| `day_name` / `month_name` | Locale-aware name lists |
| `HTMLCalendar` | Generate HTML calendar tables |

````python
from datetime import date

today = date.today()
print(today)            # 2026-06-21
print(today.year)       # 2026
print(today.month)      # 6
print(today.day)        # 21
````

the `calendar` module doesn't know "today." It only works with dates you give it.

For today's date, use the `datetime` module instead:

```python
from datetime import date

today = date.today()
print(today)            # 2026-06-21
print(today.year)       # 2026
print(today.month)      # 6
print(today.day)        # 21
```

You can combine both modules to do something useful — like printing the current month's calendar automatically:

```python
import calendar
from datetime import date

today = date.today()
print(calendar.month(today.year, today.month))
```

**Rule of thumb:**
| Need | Module |
|---|---|
| "What is today / now?" | `datetime` |
| "Show me a calendar / weekday / leap year" | `calendar` |
