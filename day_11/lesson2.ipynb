Exercise 1 — build a turn list. Start from chat = [{"role": "system", "content": "You translate English to French."}]. 
Add a user turn ("Translate: good morning") and an assistant turn ("Bonjour"), then print the number of turns. 
Use the in-place pattern correctly.

```python
chat = [{"role": "system", "content": "You translate English to French."}]

chat.append({"role": "user", "content": "what is good morning in french"})

chat.append({"role":"assistant", "content": "Bonjour"})

print(len(chat))
```
