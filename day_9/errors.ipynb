```python
butterflies = 10
beetles = 12
ladybugs = 20

total = butterflies + beetles + ladybugs

print(f'I caught {butterflies}  🦋 butterflies!')
print(f'I caught {beetles}  🪲 beetles!')
print(f'I caught {ladybugs}  🐞 ladybugs!')

print(f'I caught {str(total)} total bugs!')
```
