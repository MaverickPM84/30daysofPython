```python
# Madlibs game
#Word game where you create a story
# by filling in blanks with random words

adjective1 = input("Where did you go?: ")
noun1 = input("What did you see?: ")
adjective2 = input(f"How was the {noun1} ?: ")
verb1 = input("what did the {noun1} do?: ")
adjective3 = input("how did you feel ?:")


print(f"I went to an {adjective1} zoo.")
print(f"In a cage, I saw a {noun1}")
print(f"{noun1} was {adjective2} and {verb1}")
print(f"I was {adjective3}!")
```
