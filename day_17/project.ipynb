```python
class Pokemon:
    def __init__(self, entry_number, name, type, description, category, height, weight):
        self.entry_number = entry_number
        self.name = name
        self.type = type
        self.description = description
        self.category = category
        self.height = height
        self.weight = weight

    def speak(self):
        print(f"{self.name},{self.name} !! ")

    def display_details(self):
        print(f"Entry Number = {self.entry_number}")
        print(f"Name = {self.name}")
        print(f"Type = {self.type}")
        print(f"Description = {self.description}")

#create a Pokemon object with Pokemon Class

bulbasaur = Pokemon(1, "Bulbasaur", ["Grass", "Poison"], "For some time after its birth, it uses the nutrients that are packed into the seed on its back in order to grow.", "Seed", 2.04, 15.2 )

#call the method using object.method()

bulbasaur.speak()

#call the method using object.method()

bulbasaur.display_details()

ivysaur = Pokemon(2,"Ivysaur",["Grass", "Poison"], "The more sunlight Ivysaur bathes in, the more strength wells up within it, allowing the bud on its back to grow larger.", "Seed", 3.03, 28.7 )

print("__________________________")
print("__________________________")
ivysaur.speak()

ivysaur.display_details()

print("__________________________")
print("__________________________")

charizard = Pokemon(6, "Charizard", ["Fire", "Flying"], "Its wings can carry this Pokémon close to an altitude of 4,600 feet. It blows out fire at very high temperatures.", "Flame", 5.07, 199.5)


charizard.speak()

charizard.display_details()


```
