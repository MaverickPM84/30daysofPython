```python
# Python Calculator

operator = input("Enter an operator (+, -, *, ?):  ")

num_1 = float(input("Enter the 1st num: "))

num_2 = float(input("Enter the 2nd num: "))

if operator == "+":
    result = num_1 + num_2
    print(round(result,2))

elif operator == "-":
    result = num_1 - num_2
    print(round(result,2))

elif operator == "*":
    result = num_1 * num_2
    print(round(result,2))

elif operator == "/":
    result = num_1 / num_2
    print(round(result,2))

else:
    print("Enter a valid operator")


```
