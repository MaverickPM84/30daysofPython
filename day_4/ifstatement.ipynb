if = do some code if the condition is true
else do something else


```python
response = input("Do you want some food ? (Y/N): ")

if response == "Y":
    print("Here is some food:")
else:
    print("No food for you")

```

```python
#Boolean - use boolean values with if statement

for_sale = False

if for_sale:
    print("This item is for sale")

else:
    print("This item is not for sale")


```

```python
online = True

if online:
    print("User is Online")
else:
    print("The user is Offline")
```
