Instructions
In a five-star restaurant review system (⭐️⭐️⭐️⭐️⭐️), the stars represent different levels of satisfaction. But what does each star rating actually mean?

Start by creating a rating variable and assign it a decimal number between 0 and 5.

Next, make a rating system with an if/elif/else:

rating greater than 4.5, print 'Perfection'
rating greater than 4, print 'Excellent'
rating greater than 3, print 'Good'
rating greater than 2, print 'Fair'
Otherwise, print 'Poor'

```python
rating = 5.0

if rating > 2 and rating <= 3:
  print('Fair')
elif rating > 3 and rating <=4:
  print('Good')
elif rating > 4 and rating <= 4.5:
  print('Excellent')
elif rating > 4.5 and rating <= 5:
  print('Perfection')
else:
  print('Poor')
```

Logical operators in Python include the and and or keywords. Which one should you use?

destination weight=Earth weight × relative gravity
Number	Planet	Relative Gravity
1	Mercury	0.38

3	Mars	0.38
4	Jupiter	2.53
5	Saturn	1.07

If the user enters a planet number outside of 1 - 7, print a message that says 'Invalid number'.

```python
weight_earth = float(input("What is your weight on earth?: "))

planet_number = int(input("What is your planet's number?: "))

destination_weight = 0

if planet_number == 1:
    destination_weight = weight_earth * 0.38
    print(f'Destination weight is {destination_weight}')

elif planet_number == 2:
    destination_weight = weight_earth * 0.91
    print(f'Destination weight is {destination_weight}')

elif planet_number == 3:
    destination_weight = weight_earth * 0.38
    print(f'Destination weight is {destination_weight}')

elif planet_number == 4:
    destination_weight = weight_earth * 2.53
    print(f'Destination weight is {destination_weight}')

elif planet_number == 5:
    destination_weight = weight_earth * 1.07
    print(f'Destination weight is {destination_weight}')

elif planet_number == 6:
    destination_weight = weight_earth * 0.89
    print(f'Destination weight is {destination_weight}')

elif planet_number == 7:
    destination_weight = weight_earth * 1.14
    print(f'Destination weight is {destination_weight}')

else:
    print('Invalid number')
```
