1. Video duration formatter

A video is total_seconds = 9425 long. Using the floor-division and modulus operators, work out how many whole hours, minutes, and seconds that is, and print all three. (Expected: 2 hours, 37 minutes, 5 seconds.) This is exactly how a media player turns a raw second count into 2:37:05.

```python
hours = total_seconds // 3600
remaining_minutes = (total_seconds % 3600) // 60
remaining_seconds = total_seconds % 60

print(f"{hours}hrs, {remaining_minutes}minutes, {remaining_seconds}seconds")
```

2. Shopping cart running total

Start with cart_total = 0. A customer adds a 250-rupee item, then 3 copies of a 80-rupee item, then removes one 50-rupee item. Using only the assignment operators (+=, -=, *=, etc.), update cart_total step by step and print the final value. (Expected: 440.)

```python
cart_total = 0
cart_total += 250
cart_total += 3 * 80
cart_total -= 50
print(cart_total)
```

3. Compound interest

You invest principal = 50000 at an annual rate of 7% for 5 years, compounded yearly. Using the exponent operator, compute the final amount with the formula principal × (1 + rate) ** years, rounded to 2 decimals. (Expected: 70127.59.) The ** operator is the whole point here.

```python
principal = 50000

rate = 0.07

years = 5

final = round(principal * ( 1+ rate) ** years, 2)

print(final)
```

4. Password length check

A signup form requires passwords of at least 8 characters. Given password = "sky12", write a single expression using a comparison operator that evaluates to True if the password is long enough, and print the result. (Here it should print False.) Real validation logic starts exactly like this.

```python
password = "sky12"

result = len(password) >= 8

print(result)
```

5. Loan eligibility

A lender approves an applicant only if they are at least 21 years old AND earn more than 30000 a month. Given age = 25 and monthly_income = 28000, write one expression using a logical operator that returns whether they qualify, and print it. (Expected: False — the income fails.) Then mentally check: what would it take to flip it to True?

```python
age = 25
monthly_income = 28000
qualify = age >= 21 and monthly_income > 30000

print(qualify)
```
