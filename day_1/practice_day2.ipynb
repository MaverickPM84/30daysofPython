4. Temperature converter

A weather sensor reports temperature in Celsius as text: temp_c = "23.5". Convert it and print the Fahrenheit value rounded to 1 decimal place. (Formula: F = C × 9/5 + 32. Expected: 74.3.)

```python
temp_c = "23.5"

temp_c = float(temp_c)

temp_fahrenheit = temp_c * 9/5 + 32

print(f"(The temperature in fahrenheit is {round(temp_fahrenheit,1)})")
```

5. Quick stats on transactions

You have a day's transaction amounts:
amounts = [120.0, 45.5, 300.0, 12.99, 89.0]
Using only built-in functions (no loops), print the number of transactions, the total, the average rounded to 2 decimals, the largest, and the smallest.

```python
amounts = [120.0, 45.5, 300.0, 12.99, 89.0]

num_of_transactions = len(amounts)

print(num_of_transactions)


total = sum(amounts)

print(total)

average = round(total/num_of_transactions,2)

print(average)

largest_num = max(amounts)

print(largest_num)

smallest_num = min(amounts)

print(smallest_num)



```

6. Splitting a bill fairly

A dinner bill is 1000 rupees split between 3 people. Working in whole rupees, find how much each person pays and how many rupees are left over that don't divide evenly. There's a single built-in function that gives you both answers at once — try to find and use it. (Expected: 333 each, 1 left over.)

```python
dinner_bill = 1000

each_person = dinner_bill // 3

remainder = dinner_bill % 3

print(each_person)
print(remainder)
```

```python
#You can use the built-in Python function **`divmod()`** to get both the quotient (how much each person pays) and the remainder (what's left over) in a single step.

bill = 1000
people = 3

each_pays, leftover = divmod(bill, people)

print(f"Each person pays: {each_pays} rupees")
print(f"Left over: {leftover} rupees")


```
